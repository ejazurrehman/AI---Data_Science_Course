# What is Algebra?
---

- Algebra is a branch of mathematics that deals with symbols (variables) and the rules for manipulating them.
- Instead of working only with numbers, algebra lets us work with unknowns, relationships, and functions.
- Example:
  - 𝑦 = 2𝑥 + 5


This means if x=3, then y=11.

## Why is Algebra Important in Data Science?
---

- In data science, algebra is everywhere because we constantly model relationships between variables.
- It provides the foundation for:
### 1. Linear & Non-linear Models
  - Regression models: y=mx+c
  - Logistic regression, decision boundaries, etc.
### 2. Data Transformation
  - Scaling, normalization, feature engineering.
  - Example: 
    - 𝑧 = 𝑥 − 𝜇 / 𝜎
### 3. Optimization
  - Loss functions in ML are algebraic expressions we minimize (e.g., MSE).
### 4. Matrix Algebra (Linear Algebra)
  - Used for handling datasets with multiple features.
  - Foundation of ML algorithms (e.g., PCA, SVD, neural networks).

	​



## Practical Examples in Data Science
---

### 1. Linear Relationship (Regression Basics)

We often use algebra to model the relationship between independent (X) and dependent (Y) variables.

```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.array([1, 2, 3, 4, 5])
y = 2*x + 3   # Algebraic relationship

# Visualization
plt.scatter(x, y, color="blue", label="Data points")
plt.plot(x, y, color="red", label="y = 2x + 3")
plt.legend()
plt.title("Algebra in Linear Relationship")
plt.show()
```

**Output:** Here, the algebraic expression y=2x+3 defines the model.

### 2. Feature Scaling (Algebra in Normalization)

Algebra is used to normalize data:
𝑧 = 𝑥 − 𝜇 / 𝜎

```python
data = np.array([10, 20, 30, 40, 50])
mean, std = data.mean(), data.std()

z_scores = (data - mean) / std
print("Z-scores:", z_scores)
```

**Output:** Normalization ensures features are on the same scale for ML models.

### 3. Solving Equations for Prediction

- Suppose we have a regression model:y=5x+10
- If x=7, then y=5(7)+10=45.

```python
x = 7
y = 5*x + 10
print("Prediction:", y)
```

**Output:** Algebra helps compute predictions directly.

### 4. Matrix Algebra in ML

- For multiple features, we use matrix form: Y=X⋅β
- Where X = features, β = coefficients.

```python
# Example: 2 features
import numpy as np

X = np.array([[1, 2], [2, 3], [3, 4]])   # Features
beta = np.array([0.5, 1.5])              # Coefficients

Y = X.dot(beta)   # Linear combination
print("Predicted Y:", Y)
```

**Output:** This is how regression models work under the hood.

### Summary:
- Algebra in Data Science is the foundation of:
  - Modeling relationships (regression, classification).
  - Transforming data (scaling, standardization).
  - Optimizing loss functions.
  - Matrix algebra for multi-feature datasets and ML algorithms.

**NTOE:** Without algebra, we could not build, train, or interpret machine learning models.

## Case Study: Predicting House Prices using Algebra in Data Science
---

### Step 1. Generate a Synthetic Dataset

- We assume house prices depend on size (sqft) and number of bedrooms.
- Price = 200 × Size + 5000 × Bedrooms + 10000

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Synthetic dataset
np.random.seed(42)
size = np.random.randint(500, 3500, 50)          # house size in sqft
bedrooms = np.random.randint(1, 5, 50)           # number of bedrooms
price = 200*size + 5000*bedrooms + 10000 + np.random.randint(-10000, 10000, 50) # price with noise

df = pd.DataFrame({"Size(sqft)": size, "Bedrooms": bedrooms, "Price": price})
print(df.head())
```

### Step 2. Algebra in Data Transformation

- We normalize Size using z-score normalization:
- 𝑧 = 𝑥 − 𝜇 / 𝜎

```python
df["Size_Normalized"] = (df["Size(sqft)"] - df["Size(sqft)"].mean()) / df["Size(sqft)"].std()
print(df[["Size(sqft)", "Size_Normalized"]].head())
```

### Step 3. Algebra in Modeling (Linear Equation)

- In multiple regression:
- Price = β0 ​+ β1 ​⋅ Size + β2 ​⋅ Bedrooms
- We’ll use matrix algebra:
- Y = X ⋅ β

```python
from sklearn.linear_model import LinearRegression

# Features and target
X = df[["Size(sqft)", "Bedrooms"]]
y = df["Price"]

# Fit linear regression model
model = LinearRegression()
model.fit(X, y)

# Coefficients (β1, β2) and intercept (β0)
print("Intercept (β0):", model.intercept_)
print("Coefficients (β1, β2):", model.coef_)
```

### Step 4. Algebra in Prediction

- Using fitted coefficients, we can make predictions:
  - y(hat)​=β0​+β1​⋅x1​+β2​⋅x2​

```python
# Predict prices
df["Predicted_Price"] = model.predict(X)
print(df.head())

# Example prediction for a new house
new_house = [[2000, 3]]  # Size=2000 sqft, Bedrooms=3
predicted_price = model.predict(new_house)
print("Predicted Price for 2000 sqft, 3-bedroom house:", predicted_price[0])
```

### Step 5. Visualization

- Compare Actual vs Predicted house prices.

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Size(sqft)"], df["Price"], color="blue", label="Actual Price")
plt.scatter(df["Size(sqft)"], df["Predicted_Price"], color="red", marker="x", label="Predicted Price")
plt.xlabel("House Size (sqft)")
plt.ylabel("Price")
plt.title("Algebra in Data Science: Linear Regression")
plt.legend()
plt.show()
```

### Insights

- Algebra helped us model the relationship between size/bedrooms and price.
- Normalization used algebraic formulas to prepare features.
- Matrix algebra was used by the regression model to compute coefficients.
- Predictions were made using algebraic equations.
