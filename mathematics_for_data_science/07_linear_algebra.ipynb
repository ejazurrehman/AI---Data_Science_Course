# What is Linear Algebra?
---

Linear Algebra is a branch of mathematics that deals with vectors, matrices, and linear transformations.
It allows us to represent, manipulate, and solve problems involving large amounts of data.

Instead of dealing with single numbers (scalars), linear algebra works with:
- Scalars → single values
- Vectors → ordered list of numbers (1D)
- Matrices → 2D arrays of numbers
- Tensors → n-dimensional arrays (generalization of matrices)

## Why is Linear Algebra Important in Data Science?
---

In Data Science and ML, data is often represented as matrices & vectors:
### 1. Data Representation
- A dataset with m samples and n features is an m×n matrix.
### 2. Feature Transformation
- Scaling, rotation, and dimensionality reduction (PCA, SVD).
### 3. Machine Learning Models
- Linear regression: Y=X⋅β
- Neural networks use matrix multiplications at each layer.
### 4. Optimization
- Gradients, eigenvalues, eigenvectors.
### 5. Similarity Measures
- Dot product, cosine similarity in NLP & recommender systems.

## Key Linear Algebra Operations in Data Science
---

### 1. Representing a Dataset as a Matrix
- A dataset is usually a matrix X, where rows = samples and columns = features.

```python
import numpy as np

# Dataset: 3 samples, 2 features
X = np.array([[1, 2],
              [3, 4],
              [5, 6]])
print("Dataset (Matrix X):\n", X)
```

**NOTE:** Each row = one observation, each column = one feature.

### 2. Dot Product (Similarity)
- In ML, dot products measure similarity.
- For vectors a=[1,2], b=[3,4]:
- a⋅b=(1)(3)+(2)(4)=11

```python
a = np.array([1, 2])
b = np.array([3, 4])
dot = np.dot(a, b)
print("Dot Product:", dot)
```

### 3. Matrix Multiplication (Linear Regression)
- Linear regression formula:
- Y=X⋅β
- Where:
  - X = input features (matrix)
  - β = coefficients (vector)
  - Y = predictions (vector)

```python
# Features matrix (size, bedrooms)
X = np.array([[1000, 2],
              [1500, 3],
              [2000, 4]])

# Coefficients (price per sqft=200, price per bedroom=5000)
beta = np.array([200, 5000])

# Predictions
Y = X.dot(beta) + 10000   # adding bias (intercept)
print("Predicted Prices:", Y)
```

### 4. Eigenvalues & Eigenvectors (PCA)
- Principal Component Analysis (PCA) uses eigenvalues/eigenvectors to reduce dimensionality.

```python
from numpy.linalg import eig

# Covariance matrix
A = np.array([[2, 0],
              [0, 1]])

values, vectors = eig(A)
print("Eigenvalues:", values)
print("Eigenvectors:\n", vectors)
```

**NOTE:** Eigenvalues show importance of directions (variance), eigenvectors show directions (components).

### 5. Singular Value Decomposition (SVD)
- SVD is widely used in recommender systems (Netflix, Spotify).

```python
from numpy.linalg import svd

A = np.array([[1, 2],
              [3, 4],
              [5, 6]])

U, S, Vt = svd(A)
print("U:\n", U)
print("S:", S)
print("Vt:\n", Vt)
```
